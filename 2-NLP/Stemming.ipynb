Stemming - Stemming is the process of reducing the word to its 
word stem that affixes to suffixes and prefixes or to the roots
of words known as lemma . 

```python
words = [
    "playing",
    "played",
    "plays",
    "player",
    "running",
    "runner",
    "runs",
    "easily",
    "easy",
    "studies",
    "studying",
    "studied",
    "connection",
    "connected",
    "connecting",
    "connections",
    "happily",
    "happiness",
    "happy",
    "organization",
    "organized",
    "organizing",
    "organizations",
    "communication",
    "communicating",
    "communicated",
    "communicator"
]
```

```python
# PorterStemmer
from nltk.stem import PorterStemmer
```

```python
stemming=PorterStemmer()
```

```python
for word in words:
    print(word+" ------> "+stemming.stem(word))
```

RegexpStemmer removes parts of a word based on a regular expression (regex) pattern.

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer=RegexpStemmer('ing$|s$|e$')
```

```python
reg_stemmer.stem('eating')
```

Snowball Stemmer is a stemming algorithm provided by NLTK. It is generally more advanced than the Porter Stemmer and supports multiple languages.

```python
from nltk.stem import SnowballStemmer
```

```python
snowballStemmer=SnowballStemmer("english")
```

```python
for word in words:
     print(word+" ------> "+snowballStemmer.stem(word))
```
