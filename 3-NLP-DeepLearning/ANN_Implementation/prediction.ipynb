```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import numpy as np 
import pandas as pd
```

```python
##load the trained model , scaler pickle , onehot
from cProfile import label


model=load_model("model.h5")

#load the encoder and scaler
with open('lable_encoder_gender.pkl','rb') as file:
    lable_encoder_gender=pickle.load(file)
with open('onehot_encoder_geo.pkl','rb') as file:
    lable_encoder_geo=pickle.load(file)
with open('scaler.pkl','rb') as file:
    scaler=pickle.load(file)
```

```python
input_data={
    "EstimatedSalary":50000,
    "IsActiveMember":1,
    "HasCrCard":1,
    "NumOfProducts":2,
    "Balance":60000,
    "Tenure":3,
    "Age":40,
    "Gender":"Male",
    "CreditScore":600,
    "Geography":"France"

}
```

```python
geo_encoded=lable_encoder_geo.transform([[input_data["Geography"]]])
geo_encoded_df=pd.DataFrame(geo_encoded,columns=lable_encoder_geo.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
input_df=pd.DataFrame([input_data])
input_df
```

```python
input_df['Gender']=lable_encoder_gender.transform(input_df['Gender'])
```

```python
input_df=pd.concat([input_df.drop("Geography",axis=1),geo_encoded_df],axis=1)
input_df
```

```python
# Reorder to match the column order the scaler was fitted on
input_df = input_df[[
    'CreditScore', 'Gender', 'Age', 'Tenure', 'Balance',
    'NumOfProducts', 'HasCrCard', 'IsActiveMember', 'EstimatedSalary',
    'Geography_France', 'Geography_Germany', 'Geography_Spain'
]]
input_scaled = scaler.transform(input_df)
```

```python
predication=model.predict(input_scaled)
```
