```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sentences = [
    "Python is a popular programming language.",
    "Machine learning helps computers learn from data.",
    "Natural language processing works with human language.",
    "Text embeddings convert sentences into numerical vectors.",
    "Deep learning is a branch of machine learning.",
    "Artificial intelligence is used in many applications.",
    "Python is commonly used for machine learning.",
    "Embeddings help measure the similarity between texts.",
    "A neural network can learn patterns from data.",
    "Semantic search uses embeddings to find related text."
]
```

```python
##Define the vocabulary size
voc_size=10000
```

```python
##one hot representation
one_hot_representation=[one_hot(words,voc_size)for words in sentences]
one_hot_representation
```

```python
##word Embedding Representation
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
sent_length=8
embedded_docs=pad_sequences(one_hot_representation,padding="pre",maxlen=sent_length)
embedded_docs
```

```python
##feature representation
dim=10
```

```python
model=Sequential()
model.add(Embedding(voc_size,dim,input_length=sent_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```
